```python
# 1. Define a list of products
products = ["laptop", "mouse", "keyboard", "monitor", "headphones"]
print(products)
```

```python
# 2. Accessing items by index
message = f"Selected product: {products[0].title()}"
print(message)
```

```python
for item in products:
    print(item)
```

```python
# 4. For loop with string formatting (.title())
for item in products:
    print(f"product available: {item.title()}")
```

```python
# 5. Multi-line statements inside a loop
for item in products:
    print(f"Product: {item.title()}")
    print(f"Adding {item.title()} to your cart...")
```

```python
# 6. Adding newline delimiter ('\n') for spacing
for item in products:
    print(f"Product: {item.title()}")
    print(f"Adding {item.title()} to your cart...\n")
```

```python
# 7. Complete loop with a final message outside the loop
for item in products:
    print(f"Product: {item.title()}")
    print(f"Adding {item.title()} to your cart...\n")

print("All products have been processed successfully!")
```
