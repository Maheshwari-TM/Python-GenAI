```python
Definition: A dictionary is an unstructured collection of key-value pairs (key : value).

Mutability: Mutable (data can be modified, added, or deleted after creation).

Syntax: Declared using curly braces {} with key-value pairs separated by commas.
```

```python
alien = {'color': 'green', 'points': 5, 'level': 'stage-1'}
print(alien)
```

```python
print(alien['color'])
```

```python
print(alien['points'])
```

```python
# 1. Create a dictionary
useraccount = {
    'username': 'codetraining',
    'firstname': 'code',
    'lastname': 'training',
    'dob': '01-01-2020',
    'pwd': '54321'
}
```

```python
# 2. Access a specific key
print(useraccount['username'])
```

```python
# 3. Modify an existing value (Update password)
useraccount['pwd'] = '12345'
print(useraccount['pwd'])
```

```python
# 4. Modify another key (Update firstname)
useraccount['firstname'] = 'codet'
print(useraccount)
```

```python
del useraccount['lastname']
print(useraccount)
```
