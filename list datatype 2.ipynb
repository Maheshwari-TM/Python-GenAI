```python
# what is the diff b/w sorted andsort method in the list data type
```

```python
Both sorted() and sort() are used to arrange elements in ascending or descending order, 
but there is an important difference in how they handle the original list.

1.sorted() is a built-in function. It returns a new sorted list and does not modify the original list.
2.sort() is a method of the list. It sorts the original list in place, so the original list is modified. It returns None.     
```

```python
# Slicing of list
List slicing is used to extract a portion of elements from a list using a range of indexes. 
It does not modify the original list.
Syntax
list[start : stop : step]    
```

```python
print(companies[0:7]) 
```

```python
print(companies[2:]) 
```

```python
print(companies[:5]) #Gets the first 5 companies.
```
