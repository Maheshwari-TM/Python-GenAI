```python
#  Create List
numbers = [10, 25, 40, 25, 60, 75]

print(numbers)
```

```python
type(numbers)
```

```python
# Indexing
print(numbers[3])
print(numbers[0:5])
print(numbers[-4:])
print(numbers[-4:-1])
```

```python
#Adding

#Example 1 — append()
numbers.append(90)
print(numbers)
```

```python
#Example 2 — insert()
numbers.insert(3, 50)
print(numbers)
```

```python
# Modify
numbers[2] = 45
print(numbers)
```

```python
#Delete

#Example 1 — pop()

numbers.pop()
print(numbers)
```

```python
#Example 2 del
del numbers[4]
print(numbers)
```

```python
Interview Question: What is the difference between append() and insert() in a Python list?
```

```python
append() and insert() are both used to add an element to a Python list, but the main difference is where the element is added.

. append() adds the element at the end of the list.
    append(value) → adds at the end
. insert() adds the element at a specific index/position.
    insert(index, value) → adds at a specific position
```
