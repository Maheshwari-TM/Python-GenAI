```python
firstname = "Medari"
lastname = "Maheswari"
```

```python
# EX-1 ---> Sample
full_name = firstname + " " + lastname 
print(full_name)
print(full_name.title())
print(full_name.upper())
print(full_name.lower())
```

```python
# EX-2 ---> Using f-string
fullname = f"{firstname} {lastname}"
print(fullname)
print(fullname.title())
print(fullname.upper())
print(fullname.lower())
```

```python
# general syntax of f-strings
# EX-1 ---> Sample
message = f"My Name is {fullname.title()}"
print(message)
```

```python
# Eliminating Whitespaces ---> strip()
name1 = " Mahi"
name2 = "Mahesh "
name3 = "Maheswar"
name4 = " Maheswari "
```

```python
# Ex-1 --> left stripping
print(name1)
print(name1.lstrip())
```

```python
# Ex-2 --> right stripping
print(name2)
print(name2.rstrip())
```

```python
# Ex-3 --> Sample
print(name3)
print(name3.lstrip())
```

```python
# Ex-4 --> stripping
print(name4)
print(name4.lstrip())
```
